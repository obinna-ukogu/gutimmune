### Effect on microbiome genetics on immune repertoire

```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib import patheffects
import scipy.optimize
from scipy.optimize import minimize, LinearConstraint, Bounds, NonlinearConstraint, linprog
from math import erf
import multiprocessing as mp
# from tqdm import tqdm
from joblib import Parallel, delayed
import itertools
import os
import time

plt.style.use('/Users/ObinnaUkogu/Desktop/custom.mplstyle')
%config InlineBackend.figure_format = 'retina'
```

```python
# Define parameters
m, k = 10, 5 # (num_microbes, num_receptors)
n = 1 # Niche supply
h = 1 # Harmfulness of microbe
Sigma = np.eye(2)

rng = 50
pnts = 100
```

```python
# Define functions
def G(l, sigma = 1, n = n): # Microbe steady state population
    out = n/(1 + np.sum(l*sigma))
    return out

def objective(l, beta = 1.0, Sigma = np.eye(m), h = np.ones(m), n = np.ones(m)): # Cost function
    # beta := Cost of immune pressure

    out = np.sum([h[i]*G(l, sigma, n[i]) for i, sigma in enumerate(Sigma)]) + np.sum(beta*l)

    return out.item()
```

### Unstructured

```python
lam_opts = []
obj_opts = []
rec_num_opts = []
X_opts = []
runs = 1000

for j in np.arange(1,m + 1):
    obj_min, obj_minimizer, obj_rec_num, X = 0.0, 0.0, 0.0, 0.0
    for r in np.arange(0,runs):
        Sigma = np.random.uniform(low=0.0, high=1.0, size=(m,j))
        res_harm = minimize(objective, args = (0.1, Sigma, np.ones(m), np.ones(m)),
                 x0 = np.ones(j).tolist(),
                 tol=1e-12,
                 constraints = LinearConstraint(np.eye(j), 0.0, np.inf))
        obj_min += res_harm.fun
        obj_minimizer += res_harm.x
        obj_rec_num += np.sum(res_harm.x > 0)
        X += Sigma.dot(res_harm.x)
        
    lam_opts.append(obj_minimizer/runs)
    obj_opts.append(obj_min/runs)
    rec_num_opts.append(obj_rec_num/runs)
    X_opts.append(X/runs)
```

```python
# Plot numerical values and analytic values verus beta
fig, axs = plt.subplots(1, 4, dpi = 300, figsize=(11, 3), layout='constrained')

axs[0].scatter(np.arange(1, m + 1), rec_num_opts)
axs[0].plot([1.0,m], [1.0,m], 'k--')
axs[0].set(ylabel=r'num. unique receptors',xlabel=r'num. available receptors, $K$')

axs[1].scatter(np.arange(1, m + 1), [np.sum(u) for u in lam_opts])
axs[1].set(ylabel='total receptor\n abundance, '+r'$\sum_k^K \lambda_k$', xlabel=r'num. available receptors, $K$')

axs[2].scatter(np.arange(1, m + 1), obj_opts)
axs[2].set(ylabel=r'objective minimum, $L(\vec\lambda^*)$', xlabel=r'num. available receptors, $K$')

axs[3].scatter(np.arange(1, m + 1), [np.sum(u)/m for u in X_opts])
axs[3].set(ylabel='mean immune\n pressure,'+ r'$\frac{1}{M}\sum_\mu^M X_\mu$', xlabel=r'num. available receptors, $K$')

for j, ax in enumerate(axs.flat):
    ax.xaxis.set_tick_params(labelbottom=True)
    ax.yaxis.set_tick_params(labelleft=True)
    ax.minorticks_off()
    ax.set_box_aspect(1)
```
